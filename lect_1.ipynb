Creating a series:

               1) From list
               2) From integers
               3) Custom indexs
               4) From Dictionary

```python
country = ["Pakistan","India","Nepal","Bangladesh"]
pd.Series(country)
```

```python
integers = np.array([1,2,3,4,5])
integers = np.astype(integers,np.int16)
pd.Series(integers)
```

```python
"Custom indexes:"

indexes = ["A","B","C","D","E"]
country = ["Pakistan","india","Nepal","Bangladesh","japan"]
countries = pd.Series(country,index=indexes)
countries.name # here we have not assigne name attribute to name= so therefore we can't get name value here.
```

```python
"One more example on Custom indexs:"

marks = [90,87,76,56]
subject = ["Programming","Machine learning","Deep learning","Data science"]
student_marks_sheet = pd.Series(marks,index=subject)
student_marks_sheet
```

```python
marks = [90,87,76,56]
subject = ["Programming","Machine learning","Deep learning","Data science"]
student_marks_sheet = pd.Series(marks,index=subject,name="Waqas Marks")
student_marks_sheet
```

Series From dictionary:

Editing Series:

Using indexing:

```python
print(marks_sheet)
print()
marks_sheet[1] = 95
marks_sheet
```

```python
"""We can edit the values through two method 1) using zero base indexing and the second is label based indexing.
and the lable based indexing is the best way and not leading to any error or warning just like zero base wala.."""

print(marks_sheet)
marks_sheet["Data science"] = 95 # this will lead to permanent changes in the data...
marks_sheet
```

```python
"Normal indexing:"
run_index[0] = 50
run_index[2:4:] = 100
run_index[0::] = 10
run_index
```

```python
"Fancy indexing:"
run_index[[1,2,3]] = [0,0,0]
run_index.name = "index_range"
run_index
```

```python
"Through labeling:"
print(movies)
movies["3 Storeys"] = "Waqas Ahmad"
movies["3 Storeys"]
movies
```
