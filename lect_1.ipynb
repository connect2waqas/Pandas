Pandas Series:

```python
"What is series: A pandas series is like a column in a table. It is a 1-D array holding data of any type."
```

Creating a series:

               1) From list
               2) From integers
               3) Custom indexs
               4) From Dictionary

```python
country = ["Pakistan","India","Nepal","Bangladesh"]
pd.Series(country)
```

```python
integers = np.array([1,2,3,4,5])
integers = np.astype(integers,np.int16)
pd.Series(integers)
```

```python
"Custom indexes:"

indexes = ["A","B","C","D","E"]
country = ["Pakistan","india","Nepal","Bangladesh","japan"]
countries = pd.Series(country,index=indexes)
countries.name # here we have not assigne name attribute to name= so therefore we can't get name value here.
```

```python
"One more example on Custom indexs:"

marks = [90,87,76,56]
subject = ["Programming","Machine learning","Deep learning","Data science"]
student_marks_sheet = pd.Series(marks,index=subject)
student_marks_sheet
```

```python
marks = [90,87,76,56]
subject = ["Programming","Machine learning","Deep learning","Data science"]
student_marks_sheet = pd.Series(marks,index=subject,name="Waqas Marks")
student_marks_sheet
```

Series From dictionary:

```python
marks = {"Programming":97,
         "Machine learning":75,
         "Deep learning": 90,
         "Data science": 86}
marks_sheet = pd.Series(marks,name="Waqas marks")
print(marks_sheet.name)
print(marks_sheet)
print(marks_sheet.index)
```

```python
"""Here as we see that above sort_index() deos not make changes in the orginal 
data but if we can does this (generally not prefer) through a hidden parametar (inplace=True): (be carefull in this case.)
Here we can also use the (ascending=True or False) for getting desired result."""
movies.sort_index(inplace=True)
movies
```
