```python
def categering(value):
    if value < 12:
        return "child"
    elif 12 <= value < 18:
        return "Teen"
    elif 18 <= value < 40:
        return "Adult"
    else:
        return "Senior"

titanic["AgeGroup"] = titanic["Age"].apply(categering)
titanic
```

```python
def survivel_rate_age_group():
    titanic_survivers = titanic["Survived"].value_counts()
    series = titanic["AgeGroup"].value_counts()
    for age_group in series.index:
        rate = series[age_group] / titanic_survivers[1]
        print(f"{age_group}: {rate:.2%}")

survivel_rate_age_group()
```
