```python
"value_counts(): ---> applicable both on series and dataframe ===> more important in case of series"
marks = pd.DataFrame([
    [100,80,10],
    [90,70,8],
    [80,70,6],
    [120,100,11],
    [120,100,11]
],columns=["IQ","Marks","Package"])

"it will return the number rows repeat"
marks.value_counts()
```

```python
students = pd.DataFrame(
    {
        'name':['nitish','ankit','rupesh',np.nan,'mrityunjay',np.nan,'rishabh',np.nan,'aditya',np.nan],
        'college':['bit','iit','vit',np.nan,np.nan,'vlsi','ssit',np.nan,np.nan,'git'],
        'branch':['eee','it','cse',np.nan,'me','ce','civ','cse','bio',np.nan],
        'cgpa':[6.66,8.25,6.41,np.nan,5.6,9.0,7.4,10,7.4,np.nan],
        'package':[4,5,6,np.nan,6,7,8,9,np.nan,np.nan]
    }
)
```

```python
"Sorting on the bases of one column and througn na_postion we can tell it that nan value where to lie and inplace=True making changes permanent"
students.sort_values("name",na_position='first',ascending=False,inplace=True)
```
