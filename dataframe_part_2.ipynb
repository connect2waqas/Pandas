apply(): very important function.

```python
"when we are applying the apply() function in a series then we are actully passing a row and in case of dataframe we are passing a column to apply() ok."
points_df = pd.DataFrame(
    {
        '1st point':[(3,4),(-6,5),(0,0),(-10,1),(4,5)],
        '2nd point':[(-3,4),(0,0),(2,2),(10,10),(1,1)]
    }
)

points_df
```

```python
temp = pd.Series([10,20,30,40,50])
def sigmoid(value):
    return 1/1 + np.exp(-value)

temp.apply(sigmoid)
```

```python
points_df
```

```python
def euclidean_distance(row):
    p_A, p_B = row["1st point"], row["2nd point"]
    return ((p_A[0] - p_B[0]) **2 + (p_A[1] - p_B[1]) **2) ** 0.5
```

```python
points_df["distance"] = points_df.apply(euclidean_distance,axis=1)
```
