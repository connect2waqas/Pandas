```python
"appliying dictionary"
geners.agg(
    {
        "Runtime": "sum",
        "IMDB_Rating": "mean",
        "No_of_Votes": "sum",
        "Gross": "sum",
        "Metascore":"min"
        }
    )
```

```python
"through list:"
geners.agg(["sum","max","min"])
```

```python
"applying both syntax in one "
geners.agg(
    {
        "Runtime": ["min","max"],
        "IMDB_Rating": "mean",
        "No_of_Votes": ["sum","max"],
        "Gross": "sum",
        "Metascore":"min"
        }
    )
```

```python
"looping through groupby"
df = pd.DataFrame(columns=movies.columns)

for group, data in geners:
    df = pd.concat([df, data[data["IMDB_Rating"] == data["IMDB_Rating"].max()]])

df
```

```python
def foo(group):
    return group["Series_Title"].str.startswith("A").sum()
```

```python
geners.apply(foo)
```

```python
def rank_move(group):
    group["move_rank"] = group["IMDB_Rating"].rank(ascending=False)
    return group
```

```python
geners.apply(rank_move)
```

```python
def normal(group):
    group["normalize"] = (group["IMDB_Rating"] - group["IMDB_Rating"].min()) / (group["IMDB_Rating"].max() - group["IMDB_Rating"].min())
    return group
```

```python
geners.apply(normal)
```

```python
objects = ipl.groupby("batsman")["total_runs"].sum().sort_values(ascending=False).head(1)
objects
```

```python
highest_runs = ipl.groupby("batsman")["batsman_runs"].sum().sort_values(ascending=False).head(1)
highest_runs
```

```python
six = ipl[ipl["batsman_runs"] == 6]
```

```python
six.groupby("batsman")["batsman"].count().sort_values(ascending=False).head(1).index[0]
six
```

```python
temp_df = ipl[ipl["over"] > 15]
temp_df = temp_df[(temp_df["batsman_runs"] == 4) | (temp_df["batsman_runs"] == 6)]
temp_df.groupby("batsman")["batsman"].count().sort_values(ascending=False).head(1).index[0]
```

```python
temp_df = ipl[ipl["batsman"] == "V Kohli"]
temp_df.groupby("bowling_team")["batsman_runs"].sum().reset_index()
```

```python
def highest(batsman):
    temp_df = ipl[ipl["batsman"] == batsman]
    return temp_df.groupby("match_id")["batsman_runs"].sum().sort_values(ascending=False).head(1).values[0]
```

```python
highest("DA Warner")
```
