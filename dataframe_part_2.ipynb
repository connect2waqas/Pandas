```python
"Sorting on the bases of one column and througn na_postion we can tell it that nan value where to lie and inplace=True making changes permanent"
students.sort_values("name",na_position='first',ascending=False,inplace=True)
```

Working with Missing values:

                       1) isnull() or isna() --> series + dataframe
                       2) notnull() --> series + dataframe
                       3) hasnans() --> series + dataframe


```python
students
```

```python
students["name"].isna()
students["name"].isnull()
```

```python
"these function are both same and isnull() is basically the alias of isna()."
students.isna()
students.isnull()
```

```python
"both are same okay but notnull is alias of notna()"
students["name"].notnull()
students["name"].notna()
```

```python
"Basically (isnull() and notnull()) or (isna() and notna()) both are complementry of each other"
```

```python
"hasnans: series: (attribute)  it tell us Is there any missing value or not"
students["name"].hasnans
```
